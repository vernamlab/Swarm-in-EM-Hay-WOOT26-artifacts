```python
from npeet.entropy_estimators import mi, micd, midd, midc
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
R = np.array(pd.read_csv('X.csv', header=None))
MBREs = np.array(pd.read_csv('mi_vals_r.csv', header=None)).transpose()
ACCs = np.array(pd.read_csv('mi_vals.csv', header=None)).transpose()
```

```python
def KSG_MI(x,y):

    o = np.array(x, dtype=int)
    h = np.array(y, dtype=int)

    
    return mi(o.reshape(-1, 1) , h.reshape(-1, 1))
```

```python


import matplotlib.pyplot as plt
X = R[:,0]
Y=R[:,1]


X = R[:,0]
Y=R[:,1]
sigmas =np.linspace(0, 6, 200)

KSGs = []

for i in range(len(sigmas)):
    sigma = sigmas[i]
    noise = sigma * np.random.randn(700, 2)


    KSGs.append(KSG_MI(X+noise[:,0],Y+noise[:,1]))
    
    
x = sigmas
ax = plt.figure(figsize=(7.5, 6))
plt.plot(x, KSGs, label= "KSG"  , color = 'green')
plt.plot(x, MBREs, label= "MBRE", color = 'orange')
plt.plot(x, ACCs,label= "Analytical", color = 'blue')
plt.xlabel('sigma', fontsize=20)  # Adjust font size as needed
plt.ylabel('MI', fontsize=20)
plt.xlabel(f'$\sigma$(noise)')
plt.ylabel("MI")
plt.tick_params(axis='both', labelsize=16)  # You can change 12 to your desired size
plt.xlim(0,6)

plt.legend(fontsize=16)
```
